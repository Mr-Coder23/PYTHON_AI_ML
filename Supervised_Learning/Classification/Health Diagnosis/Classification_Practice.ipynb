```python
import pandas as pd
import numpy as np
print("Data Upload : ")
df=pd.read_csv("heart.csv")
data=pd.DataFrame(df)
print(data.shape)
print(data.info())
```

```python
print(data.isnull().sum())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn import svm
from sklearn.metrics import accuracy_score,recall_score,precision_score,f1_score,confusion_matrix
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
X=data.drop("target",axis=1)
y=data["target"]
print(X.shape)
print(y.shape)
print(X.columns)
print(y.name)
```

```python
X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=0.4,random_state=10)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(X_test.shape)
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)

print(X_train_scaled[:2])
```

```python
model1=LogisticRegression()
model2=KNeighborsClassifier(n_neighbors=5)
model3=svm.SVC(kernel="rbf",gamma="scale",C=10)
model4=svm.SVC(kernel="poly",gamma="scale",C=100)
model5=svm.SVC(kernel="linear",gamma="scale",C=1000)
print("All Models : ")
print(model1)
print(model2)
print(model3)
print(model4)
print(model5)
```

```python
print("Fitting,Predicting and Evaluating of Model 1 : ")
model1.fit(X_train_scaled,y_train)
y1=model1.predict(X_test_scaled)
print("Results of Model 1 : \n",y1)
print("--------------------------------------------\nEvaluation of Model 1 (Logistic Regression):\n--------------------------------------------")
Accuracy1=accuracy_score(y_test,y1)
Precision1=precision_score(y_test,y1)
Recall1=recall_score(y_test,y1)
f1_1=f1_score(y_test,y1)
print("Accuracy : ",Accuracy1)
print("Precision : ",Precision1)
print("Recall : ",Recall1)
print("F1 Score : "  ,f1_1)
cm1=confusion_matrix(y_test,y1)
print("Confusion Matrix : ")
print(cm1)
```

```python
print("Fitting,Predicting and Evaluating of Model 2 : ")
model2.fit(X_train_scaled,y_train)
y2=model2.predict(X_test_scaled)
print("Results of Model 2 : \n",y2)
print("--------------------------------------------\nEvaluation of Model 2 (KNeighborsClassifier):\n--------------------------------------------")
Accuracy2=accuracy_score(y_test,y2)
Precision2=precision_score(y_test,y2)
Recall2=recall_score(y_test,y2)
f1_2=f1_score(y_test,y2)
print("Accuracy : ",Accuracy2)
print("Precision : ",Precision2)
print("Recall : ",Recall2)
print("F1 Score : "  ,f1_2)
cm2=confusion_matrix(y_test,y2)
print("Confusion Matrix : ")
print(cm2)
```

```python
print("Fitting,Predicting and Evaluating of Model 2.1 : (KNN Hyperparameter Tuning) ")
param_grid={'n_neighbors':[1,3,5,7,9,11]}
grid_search=GridSearchCV(KNeighborsClassifier(),param_grid,cv=5)
grid_search.fit(X_train_scaled,y_train)
best_knn=grid_search.best_estimator_
print(best_knn)
y21=best_knn.predict(X_test_scaled)
print("Results of Tuned KNN : \n",y21)
print("Results of Model 2 : \n",y2)
print("--------------------------------------------\nEvaluation of Model 2.1 (KNeighborsClassifier):\n--------------------------------------------")
Accuracy21=accuracy_score(y_test,y21)
Precision21=precision_score(y_test,y21)
Recall21=recall_score(y_test,y21)
f1_21=f1_score(y_test,y21)
print("Accuracy : ",Accuracy21)
print("Precision : ",Precision21)
print("Recall : ",Recall21)
print("F1 Score : "  ,f1_21)
cm21=confusion_matrix(y_test,y21)
print("Confusion Matrix : ")
print(cm21)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

models = {
    "Logistic Regression": LogisticRegression(),
    "KNN (k=5)": KNeighborsClassifier(n_neighbors=5),
    "KNN (k=9)": KNeighborsClassifier(n_neighbors=9),
    "SVM-RBF": SVC(kernel="rbf", gamma="scale", C=100),
    "SVM-Polynomial": SVC(kernel="poly", gamma="scale", C=100),
    "SVM-Linear": SVC(kernel="linear", C=100)
}

results = []

for name, model in models.items():

    # Fitting
    model.fit(X_train_scaled, y_train)

    # Prediction
    y_pred = model.predict(X_test_scaled)

    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)

    TN, FP, FN, TP = cm.ravel()

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    results.append([
        name,
        accuracy,
        precision,
        recall,
        f1,
        TN,
        FP,
        FN,
        TP
    ])

# Create comparison DataFrame
comparison = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "TN",
        "FP",
        "FN",
        "TP"
    ]
)

print(comparison)
```

```python
from sklearn.model_selection import cross_val_score

models = {
    "Logistic Regression": LogisticRegression(),
    "KNN (k=5)": KNeighborsClassifier(n_neighbors=5),
    "KNN (k=9)": KNeighborsClassifier(n_neighbors=9),
    "SVM-RBF": SVC(kernel="rbf", gamma="scale", C=100),
    "SVM-Polynomial": SVC(kernel="poly", gamma="scale", C=100),
    "SVM-Linear": SVC(kernel="linear", C=100)
}

for name, model in models.items():

    scores = cross_val_score(
        model,
        X_train_scaled,
        y_train,
        cv=5,
        scoring="accuracy"
    )

    print(name)
    print("CV Scores:", scores)
    print("Mean Accuracy:", scores.mean())
    print("Standard Deviation:", scores.std())
    print("--------------------------------||||")
```
