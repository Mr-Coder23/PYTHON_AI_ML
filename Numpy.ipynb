```python
import pandas as pd

data = {
    "SquareFeet": [1000, 1200, 1500, 1800, 2000, 2200, 2500, 3000],
    "Bedrooms": [2, 2, 3, 3, 4, 4, 4, 5],
    "Price": [150000, 180000, 250000, 310000, 380000, 420000, 500000, 620000]
}

df = pd.DataFrame(data)

print(df)
```

```python
X=df[["SquareFeet","Bedrooms"]]
y=df["Price"]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42)
print("X_train:\n",X_train)
print("X_test:\n",X_test)
print("Y_train:\n",y_train)
print("Y_test:\n",y_test)
    
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
print(y_pred)
```

```python
print(model.coef_)
print(model.intercept_)
```

```python
import matplotlib.pyplot as plt
def plot_regression(x,y):
    plt.figure(figsize=(8,6))
    plt.scatter(x,y,color="blue",label="Data Points")
    plt.xlabel("Square Feet")
    plt.ylabel("Price")
    plt.title("Hosuing Price Dataset")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
df=pd.DataFrame(data)
print(df)
plot_regression(df["SquareFeet"],df["Price"])
```
