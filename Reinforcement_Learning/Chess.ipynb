```python
import chess

env = ChessEnv()
env.reset()

env.render()

while not env.board.is_game_over():

    # First coordinate
    from_square = input("Enter starting position (e.g. e2): ")

    # Second coordinate
    to_square = input("Enter destination position (e.g. e4): ")

    # Convert coordinates into a chess move
    move = chess.Move.from_uci(from_square + to_square)

    # Check whether move is legal
    if move in env.board.legal_moves:

        state, reward, done = env.step(move)

        print("\nMove:", from_square, "->", to_square)
        print("Reward:", reward)
        print("Game Over:", done)

        env.render()

    else:
        print("❌ Invalid chess move! Try again.")
```
