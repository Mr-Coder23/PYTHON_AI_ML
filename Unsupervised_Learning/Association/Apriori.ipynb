```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules


# --------------------------------------------------
# 1. Create Dataset
# --------------------------------------------------

data = {
    'TID': [1,1,1,2,2,3,3,3,4,4,4,5,5,1,2],
    'ITEM': [
        'Milk', 'Bread', 'Diaper',
        'Coke', 'Milk',
        'Bread', 'Diaper', 'Diaper',
        'Coke', 'Bread', 'Bread',
        'Coke', 'Diaper',
        'Milk', 'Diaper'
    ]
}

print("Original Data:")
print(data)


# --------------------------------------------------
# 2. Convert Dictionary into DataFrame
# --------------------------------------------------

df = pd.DataFrame(data)

print("\nDataFrame:")
print(df)


# --------------------------------------------------
# 3. Create Basket
# --------------------------------------------------

basket = (
    df.groupby(['TID', 'ITEM'])['ITEM']
      .count()
      .unstack()
      .reset_index()
      .fillna(0)
      .set_index('TID')
)

print("\nBasket:")
print(basket)


# --------------------------------------------------
# 4. Convert Basket into Boolean Values
# --------------------------------------------------

basket = basket.astype(bool)

print("\nBoolean Basket:")
print(basket)


# --------------------------------------------------
# 5. Apply Apriori
# --------------------------------------------------

frequent_itemsets = apriori(
    basket,
    min_support=0.2,
    use_colnames=True
)

print("\nFrequent Itemsets:")
print(frequent_itemsets)


# --------------------------------------------------
# 6. Generate Association Rules
# --------------------------------------------------

rules = association_rules(
    frequent_itemsets,
    metric='confidence',
    min_threshold=0.6
)

print("\nAssociation Rules:")
print(rules)
```
