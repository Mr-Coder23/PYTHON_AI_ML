```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
from itertools import combinations

# Sample Transactional Data
data = {
    'TID': [1,1,1,2,2,3,3,3,4,4,4,5,5,1,2],
    'ITEM': [
        'Milk', 'Bread', 'Diaper',
        'Coke', 'Milk',
        'Bread', 'Diaper', 'Diaper',
        'Coke', 'Bread', 'Bread',
        'Coke', 'Diaper',
        'Milk', 'Diaper'
    ]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)


# --------------------------------------------------
# STEP 1: Convert Horizontal Transaction Data
# --------------------------------------------------

transactions = df.groupby('TID')['ITEM'].apply(list).tolist()

print("\nTransactions:")
print(transactions)


# --------------------------------------------------
# STEP 2: Convert to Vertical Data Format
# --------------------------------------------------

def create_vertical_data(transactions):

    vertical_data = {}

    for tid, transaction in enumerate(transactions, start=1):

        for item in transaction:

            if item not in vertical_data:
                vertical_data[item] = set()

            vertical_data[item].add(tid)

    return vertical_data


vertical_data = create_vertical_data(transactions)

print("\nVertical Data:")
for item, tids in vertical_data.items():
    print(item, ":", tids)


# --------------------------------------------------
# STEP 3: Calculate Support
# --------------------------------------------------

def calculate_support(itemset, vertical_data, num_transactions):

    if isinstance(itemset, str):

        itemset_tids = vertical_data.get(itemset, set())

    else:

        tid_sets = [
            vertical_data.get(item, set())
            for item in itemset
        ]

        itemset_tids = set.intersection(*tid_sets)

    support = len(itemset_tids) / num_transactions

    return support


# --------------------------------------------------
# STEP 4: Minimum Support
# --------------------------------------------------

min_support = 0.5

num_transactions = df['TID'].nunique()

print("\nNumber of Transactions:", num_transactions)
print("Minimum Support:", min_support)


# --------------------------------------------------
# STEP 5: Frequent 1-Itemsets
# --------------------------------------------------

frequent_1_itemsets = {}

for item in vertical_data:

    support = calculate_support(
        item,
        vertical_data,
        num_transactions
    )

    if support >= min_support:

        frequent_1_itemsets[item] = support


print("\nFrequent 1-Itemsets:")

for item, support in frequent_1_itemsets.items():

    print(
        item,
        "-> Support:",
        support
    )


# --------------------------------------------------
# STEP 6: Frequent 2-Itemsets
# --------------------------------------------------

frequent_2_itemsets = {}

for item1, item2 in combinations(
    frequent_1_itemsets.keys(), 2
):

    itemset = (item1, item2)

    support = calculate_support(
        itemset,
        vertical_data,
        num_transactions
    )

    if support >= min_support:

        frequent_2_itemsets[itemset] = support


print("\nFrequent 2-Itemsets:")

for itemset, support in frequent_2_itemsets.items():

    print(
        itemset,
        "-> Support:",
        support
    )


# --------------------------------------------------
# STEP 7: Frequent 3-Itemsets
# --------------------------------------------------

frequent_3_itemsets = {}

for itemset in combinations(
    frequent_1_itemsets.keys(), 3
):

    support = calculate_support(
        itemset,
        vertical_data,
        num_transactions
    )

    if support >= min_support:

        frequent_3_itemsets[itemset] = support


print("\nFrequent 3-Itemsets:")

for itemset, support in frequent_3_itemsets.items():

    print(
        itemset,
        "-> Support:",
        support
    )
```
