```python
import tensorflow as tf
import numpy as np
import os
#Load CIFAR-10 
(x_train,y_train),(x_test,y_test)=tf.keras.datasets.cifar10.load_data()
print("Training Images : " , x_train.shape)
print("Training Labels : " , y_train.shape)


```

```python
class_names=[
    "airplane",
    "automobile",
    "bird",
    "cat",
    "deer",
    "dog",
    "frog",
    "horse",
    "ship",
    "truck"
]
print(class_names)
```

```python
base_dir="CNN"
os.makedirs(os.path.join(base_dir,"Bird"),exist_ok=True)
os.makedirs(os.path.join(base_dir,"Cat"),exist_ok=True)
os.makedirs(os.path.join(base_dir,"Dog"),exist_ok=True)
print("Folders Created!")
```

```python
classes={"Bird" : 2,
         "Cat" : 3,
         "Dog" : 5}
for class_name, class_id in classes.items():
    #Fina all images belonging to this class
    indices=np.where(y_train.flatten()==class_id)[0]
    #Take first 200
    selected_indices=indices[:200]
    #Save Images
    for i, index in enumerate(selected_indices):
        image=x_train[index]
        image=Image.fromarray(image)
        image_path=os.path.join(
            base_dir,
            class_name,
            f"{class_name.lower()}_{i+1:03d}.png"
        )
        image.save(image_path)
    print(f"{class_name}: {len(selected_indices)} images saved")
```

```python
import os
from PIL import Image
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))

classes = ["Bird", "Cat", "Dog"]

for i, class_name in enumerate(classes):

    folder = os.path.join(base_dir, class_name)

    # Only select actual image files
    image_files = [
        f for f in os.listdir(folder)
        if f.lower().endswith((".png", ".jpg", ".jpeg"))
    ]

    print(class_name, ":", len(image_files), "images")

    # Take the first image
    image_path = os.path.join(folder, image_files[0])

    img = Image.open(image_path)

    plt.subplot(1, 3, i + 1)
    plt.imshow(img)
    plt.title(class_name)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
