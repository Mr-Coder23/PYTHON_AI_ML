```python
from tensorflow.keras.datasets import imdb
(train_data,train_labels),_=imdb.load_data(num_words=1000)
print(train_data.shape)
print(train_labels.shape)
print("First Review : ", train_data[0])
```

```python
import matplotlib.pyplot as plt

review_lengths = [len(review) for review in train_data]

plt.hist(review_lengths, bins=30)
plt.xlabel("Number of Words")
plt.ylabel("Number of Reviews")
plt.title("Distribution of IMDB Review Lengths")
plt.show()
```

```python
print("Minimum length:", min(review_lengths))
print("Maximum length:", max(review_lengths))
print("Average length:", sum(review_lengths) / len(review_lengths))
```

```python
import numpy as np

print("Minimum:", np.min(review_lengths))
print("Maximum:", np.max(review_lengths))
print("Mean:", np.mean(review_lengths))
print("Median:", np.median(review_lengths))
```
