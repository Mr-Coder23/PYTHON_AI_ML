```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

weighted_sum_pos = 2
output_pos = sigmoid(weighted_sum_pos)

weighted_sum_neg = -2
output_neg = sigmoid(weighted_sum_neg)

print(f"Output for weighted sum of 2: {output_pos:.2f}")
print(f"Output for weighted sum of -2: {output_neg:.2f}")
```
