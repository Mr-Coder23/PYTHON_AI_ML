```python
sentence1="The cat sat on the mat."
sentence2="The dog sat on the cat."
tokens1=sentence1.lower().split()
tokens2=sentence2.lower().split()
print("Tokens 1 : ",tokens1)
print("Tokens 2 : " , tokens2)
```

```python
text="This is an example sentence."
tokens=text.split()
print(tokens)
print("Lower case tokens : " , text.lower())
```

```python
import string
text="Hello, World!"
punctuation=string.punctuation
print(punctuation)
no_punct=''.join(char for char in text if char not in punctuation)
print(no_punct)
```
