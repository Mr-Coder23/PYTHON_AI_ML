```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
text="The quick brown fox jumps over a lazy dog"
print("Stemmatization Example : ")
print("Original Text : ",text)
words=word_tokenize(text.lower())
stemmer=PorterStemmer()
stemmed_words=[stemmer.stem(word) for word in words]
print(stemmed_words)
```

```python
print("Lemmatization Example :" )
lemmatizer=WordNetLemmatizer()
lemmatized_words=[lemmatizer.lemmatize(word) for word in words]
print("Lemamtized Vocabulary : \n", lemmatized_words)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
documents=[
    "This is the first document.",
    "This document is the second document.",
    "And this is the third one.",
    "Is this the first document?"]
vectorizer=CountVectorizer()
X=vectorizer.fit_transform(documents)
print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
import pandas as pd
from sklearn.naive_bayes import MultinomialNB
data=[
    ("Free Money!","spam"),("Meet me tonight.","not spam"),("Claim your prize!","spam"),("See you later.","not spam")
]
df=pd.DataFrame(data)
print(df)
texts,labels=zip(*data)
vectorizer=CountVectorizer()
X=vectorizer.fit_transform(texts)
model=MultinomialNB()
model.fit(X,labels)
```

```python
new_texts=["Claim your prize","How are you?"]
new_X=vectorizer.transform(new_texts)
predictions=model.predict(new_X)
for text, prediction in zip(new_texts, predictions):
    print(text,"->",prediction)
```
