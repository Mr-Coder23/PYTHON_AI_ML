```python
import numpy as np
```

```python
print("1D Array:")
arr1=np.array([1,2,3,4])
print(arr1)
```

```python
print("2D Array:")
arr2=np.array([[1,2],[3,4]])
print(arr2)
```

```python
print("Array1 : " ,arr1)
print("Dimension of arr1 : ",arr1.ndim)
print("Size of arr1: " ,arr1.size)
print("Array2 : \n" ,arr2)
print("Dimension of arr2 : ",arr2.ndim)
print("Size of arr2 : " ,arr2.size)
print("Data types : ",arr1.dtype,arr2.dtype)
```

```python
print("Special Arrays : ")
print("Arrays of 0\n",np.zeros((2,3)))
print("Arrays of 1\n",np.ones((3,3)))
print("Identity Matrix\n",np.eye(3))
print("Arange 1 : ",np.arange(1,11))
print("Arange 2 : ",np.arange(1,11,2))
print("Arange 3 : ",np.arange(12))
print("Linspace 1 \n: ",np.linspace(0,10))
print("Linspace 2 \n: ",np.linspace(0,10,5))
```

```python
print("Reshaping Array")
b=np.arange(10)
print(f"New Array : \n{b}")
print("The array in the dimension of 2*5 : ")
print(b.reshape(2,5))
print("The array in the dimension of 5*2 : ")
print(b.reshape(5,2))
```

```python
print("Indexing")
print("Array :",b)
print("Index 0 : ",b[0])
print("Index 1 : ",b[1])
```

```python
print("1-D Slicing")
print(b[1:4])
print("2-D Slicing")
c=np.array([[1,2,3],[2,3,4],[4,6,7]])
print("New 2-D Array :\n",c)
print(c[0:2,1:3])
```

```python
print("Arithmetic Operations : ")
p=np.array([1,2,3])
q=np.array([4,5,6])
print(f"Addition of Array p and Array q\n{p} + {q} =",p+q)
print(f"Subtraction of Array p and Array q\n{p} - {q} =",p-q)
print(f"Multiplication of Array p and Array q\n{p} * {q} =",p*q)
print(f"Division of Array p and Array q\n{p} / {q} =",p/q)
print("\nScalar Operations : ")
k=int(input("Enter an integer"))
print(f"{p} + {k} =",p+k)
```

```python
print("Aggregate Function")
a=np.array([10,20,30,40])
print("Sum of array : ",np.sum(a))
print("Mean of array : ",np.mean(a))
print("Max of array : ",np.max(a))
print("Min of array : ",np.min(a))
print("Standard Deviation of array : ",np.std(a))
```

```python
print("Matrix Operation " )
a=np.array([[1,2],[3,4]])
b=np.array([[2,3],[4,5]])
print("a:\n",a)
print("b:\n",b)
# . and @ are used for dot product
print(f"a dot b:\n=",np.dot(a,b))
print(f"a @ b=\n",a@b)
print("Transpose :\n",a.T)
```

```python
x=np.arange(10,20,5)
print(x[0])
y=np.array([1,2,3])
print(y<x[0])
```

```python
print('Universal Functions')
a=np.array([1,4,9,16])
print(np.sqrt(a))
print(np.square(a))
print(np.sin(a))
print(np.exp(a))
print(np.log(a))
```

```python
print("Sorting")
d=np.array([23,21,34,43,56,12])
print(d)
print(np.sort(a))
print('Concatenation')
a=np.array([1,2])
b=np.array([3,4])
print(np.concatenate((a,b)))
```

```python
print("Splitting")
p=np.arange(1,11)
print(p)
print(np.split(p,2))
```

```python
print("Random Numbers.\nRandom Numbers in points\n",np.random.rand(3))
print("A random number under 5 : ",np.random.randint(5))
print("List of 5 random numbers under 10 : ",np.random.randint(1,10,5))
```

```python
a=np.array([1,2,3,4])
b=a.copy()
c=a.view()
a[0]=100
print(a)
print(b)
print(c)
```

```python
print(np.argmax(a))

# Index of minimum
print(np.argmin(a))
# Unique elements
print(np.unique(a))

# Flatten array
print(a.flatten())

# Reshape
print(a.reshape(2,2))

# Transpose
print(a.reshape(2,2).T)
```
